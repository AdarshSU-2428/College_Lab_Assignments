```python
# Q1: Write a function that returns the sum of digits of a number using recursion.
def sum_digits(n):
    if n == 0:
        return 0
    return n % 10 + sum_digits(n // 10)

n = int(input("Enter a number: "))
print("Sum of digits:", sum_digits(n))
```

```python
# Q2: Write a function to check whether a number is an Armstrong number (without using power function).
def power(a, b):
    if b == 0:
        return 1
    return a * power(a, b - 1)

def is_armstrong(n):
    temp = n
    digits = len(str(n))
    s = 0
    while temp > 0:
        d = temp % 10
        s += power(d, digits)
        temp //= 10
    return s == n

n = int(input("Enter number: "))
print("Armstrong:", is_armstrong(n))
```

```python
# Q3: Write a function that returns the largest digit in a number.
def largest_digit(n):
    max_d = 0
    while n > 0:
        d = n % 10
        if d > max_d:
            max_d = d
        n //= 10
    return max_d

n = int(input("Enter number: "))
print("Largest digit:", largest_digit(n))
```

```python
# Q4: Write a function to check whether a number is palindrome.
def is_palindrome(n):
    rev, temp = 0, n
    while temp > 0:
        rev = rev * 10 + temp % 10
        temp //= 10
    return rev == n

n = int(input("Enter number: "))
print("Palindrome:", is_palindrome(n))
```

```python
# Q5: Write a function that returns count of prime digits.
def count_prime_digits(n):
    primes = [2, 3, 5, 7]
    count = 0
    while n > 0:
        if n % 10 in primes:
            count += 1
        n //= 10
    return count

n = int(input("Enter number: "))
print("Prime digits count:", count_prime_digits(n))
```

```python
# Q6: Write a function to rotate digits of a number k times to the right.
def rotate(n, k):
    s = str(n)
    k = k % len(s)
    return int(s[-k:] + s[:-k])

n = int(input("Enter number: "))
k = int(input("Enter rotations: "))
print("Rotated number:", rotate(n, k))
```

```python
# Q7: Write a function to find factorial using recursion.
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

n = int(input("Enter number: "))
print("Factorial:", factorial(n))
```

```python
# Q8: Write a function to check whether a number is perfect.
def is_perfect(n):
    s = 0
    for i in range(1, n):
        if n % i == 0:
            s += i
    return s == n

n = int(input("Enter number: "))
print("Perfect number:", is_perfect(n))
```

```python
# Q9: Write a function to find GCD of two numbers using recursion.
def gcd(a, b):
    if b == 0:
        return a
    return gcd(b, a % b)

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
print("GCD:", gcd(a, b))
```

```python
# Q10: Write a function to find LCM using GCD.
def gcd(a, b):
    if b == 0:
        return a
    return gcd(b, a % b)

def lcm(a, b):
    return (a * b) // gcd(a, b)

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
print("LCM:", lcm(a, b))
```

```python
# Q11: Find second largest element without using sort() or max().
def second_largest(lst):
    first = second = float('-inf')
    for x in lst:
        if x > first:
            second = first
            first = x
        elif x > second and x != first:
            second = x
    return second

lst = list(map(int, input("Enter list: ").split()))
print("Second largest:", second_largest(lst))
```

```python
# Q12: Count number of vowels in a string.
def count_vowels(s):
    count = 0
    for ch in s.lower():
        if ch in "aeiou":
            count += 1
    return count

s = input("Enter string: ")
print("Vowel count:", count_vowels(s))
```

```python
# Q13: Remove duplicate elements from a list.
def remove_duplicates(lst):
    result = []
    for x in lst:
        if x not in result:
            result.append(x)
    return result

lst = list(map(int, input("Enter list: ").split()))
print("Without duplicates:", remove_duplicates(lst))
```

```python
# Q14: Find frequency of each character in a string.
def freq(s):
    d = {}
    for ch in s:
        d[ch] = d.get(ch, 0) + 1
    return d

s = input("Enter string: ")
print("Frequency:", freq(s))
```

```python
# Q15: Check whether two strings are anagrams.
def freq(s):
    d = {}
    for ch in s:
        d[ch] = d.get(ch, 0) + 1
    return d

def is_anagram(s1, s2):
    return freq(s1) == freq(s2)

s1 = input("Enter first string: ")
s2 = input("Enter second string: ")
print("Anagram:", is_anagram(s1, s2))
```

```python
# Q16: Sum of factorial of digits.
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

def sum_fact_digits(n):
    s = 0
    while n > 0:
        s += factorial(n % 10)
        n //= 10
    return s

n = int(input("Enter number: "))
print(sum_fact_digits(n))
```

```python
# Q17: Find nth Fibonacci using recursion.
def fib(n):
    if n <= 1:
        return n
    return fib(n-1) + fib(n-2)

n = int(input("Enter n: "))
print(fib(n))
```

```python
# Q18: Find next greater number using same digits.
def next_greater(n):
    digits = list(str(n))
    i = len(digits) - 2
    while i >= 0 and digits[i] >= digits[i+1]:
        i -= 1
    if i == -1:
        return n
    j = len(digits) - 1
    while digits[j] <= digits[i]:
        j -= 1
    digits[i], digits[j] = digits[j], digits[i]
    digits[i+1:] = reversed(digits[i+1:])
    return int("".join(digits))

n = int(input())
print(next_greater(n))
```
