```python
# 1. Write a program to count the number of times a character appears in a given string.
s = input("Enter a string: ")
freq = {}
for ch in s :
  if ch in s:
    x = s.count(ch)
  freq[ch] = x

print(freq)
```

```python
# 6. Write a program that checks if two same values in a dictionary have different keys. That is, for dictionary D1 = { 'a' : 10, 'b': 20, 'c' : 10}, the program should print 2 keys have same values and for dictionary D2 = {'a' : 10, 'b' : 20, 'c' : 30} , the program should print No keys have same values.
d1 = {'a' : 10, 'b': 20, 'c' : 10}
d2 = {'a' : 10, 'b' : 20, 'c' : 30}

found = False
seen = {}

for value in d1.values():
    seen[value] = seen.get(value, 0) + 1

for key, value in seen.items():
    if value > 1:
        if not found:
            print("No. of keys with same value are:")

        found = True
        print("Value", value, "repeated", key, "times")

if not found:
    print("No key have repeated values")
```

```python
# 7. Write a program to check if a dictionary is contained in another dictionary e.g., if d1 = {1:11, 2:12} d2 = {1:11, 2:12, 3:13, 4:15} then d1 is contained in d2.
d1 = {1:11, 2:12}
d2 = {1:11, 2:12, 3:13, 4:15}
flag = True

for key, value in d1.items():
    if key not in d2 or d2[key] != value:
        flag = False

if flag:
    print("d2 contains d1")
```

```python
# 8. A dictionary D1 has values in the form of lists of numbers. Write a program to create a new dictionary D2 having same keys as D1 but values as the sum of the list elements e.g., D1 = {'A' : [1, 2, 3] , 'B' : [4, 5, 6]} then D2 is {'A' :6, 'B' : 15}
d1 = {'A' : [1, 2, 3] , 'B' : [4, 5, 6]}
d2 = {key: sum(value) for key, value in d1.items()}
print("Resultant dict:", d2)
```

```python
# 13. Invert Dictionary with Duplicate Value
  # Input: d = {'a':1, 'b':2, 'c':1, 'd':3, 'e':2}
  # Output: {1: ['a', 'c'], 2: ['b', 'e'], 3: ['d']}

d = {'a':1, 'b':2, 'c':1, 'd':3, 'e':2}
res = {}

for key, value in d.items():
    res[value] = res.get(value, [])
    res[value].append(key)

print("resultant dict:", res)
```

```python
# 14. Using dictionary, print the first character that does not repeat.
  # Input: "aabbccddeefg"
  # Output: f

string = input("Enter a string: ")
freq = {}

for ch in string:
    freq[ch] = freq.get(ch, 0) + 1

for ch in string:
    if freq[ch] == 1:
        print("The first character that does not repeat:", ch)
        break
```

```python
# 15. Using dictionary, count how many times each number appears.
nums = input("Enter the numbers").split()
lst = [int(num) for num in nums]
freq = {}

for num in lst:
    freq[num] = freq.get(num, 0) + 1

print("Frequency of each number is:", freq)
```

```python
# Q16. Create a dictionary where keys are word lengths and values are lists of words.
      # Input: ["cat","dog","elephant","rat","tiger"]
      # Output: {3:['cat','dog','rat'], 5:['tiger'],8:['elephant']}

words = list(eval(input("Enter the list of strings: ")))
dict1 = {}

for word in words:
    dict1[len(word)] = dict1.get(len(word), [])
    dict1[len(word)].append(word)

print("Resultant dict is:", dict1)
```
