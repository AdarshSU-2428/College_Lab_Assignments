```python
# Write a program to find the smallest and largest digit in a number.
n = int(input("Enter a number: "))
max = 0
min = 9

for i in str(n) :
  digit = n % 10
  if digit < min:
    min = digit
  if digit > max :
    max = digit
  n //= 10

print("The largest digit is: ", max)
print("The smallest digit is: ", min)
```

```python
# Write a program to count how many digits are even and odd in a number.
n = int(input("Enter a number: "))
even, odd = 0, 0
for i in str(n) :
  digit = n % 10
  if digit % 2 == 0 :
    even += 1
  else :
    odd += 1
  n //= 10

print("Even: ", even)
print("Odd: ", odd)
```

```python
# Count numbers between 1 and 100 whose reverse is greater than the original number.
count = 0
for i in range(1, 101) :
  rev, temp = 0, i
  while temp > 0 :
    rev = rev * 10 + (i % 10)
    temp //= 10

  if rev > i :
    count += 1

print(count)
```

```python
# Print all numbers from 1 to 100 that are not divisible by any digit they contain.
for i in range(1, 101) :
  temp = i
  for j in str(temp) :
    digit = temp % 10
    if digit and i % digit == 0 :
      break

    temp //= 10
  else :
    print(i, end=" ")
```

```python
# Check whether a number is a Harshad number (divisible by sum of digits).
n = int(input("Enter a number: "))
sum, temp = 0, n
while temp > 0:
  sum += temp % 10
  temp //= 10
if n % sum == 0:
  print("Harshad number")
else :
  print("Not a harshad number")
```

```python
# Find numbers whose square ends with the same digits as the number itself.(Hit: 5->5^2=25)
low = int(input("Enter the low number: "))
high = int(input("Enter the high number: "))
for i in range(low, high+1) :
  squared, temp = i**2, i
  flag = True
  while temp > 0 :
    if squared % 10 != temp % 10 :
      flag = False
      break
    temp //= 10
    squared //= 10

  if flag:
    print(i, end=" ")
```

```python
# Find numbers between 1 and 1000 that become prime after removing one digit.
for i in range(1, 1001) :
  factor, temp = 1, i
  while temp :
    num = (i // (10 * factor)) * factor + (i % factor)

    if num > 1 :
      flag = True
      for j in range(2, num) :
        if num % j == 0 :
          flag = False
          break

      if flag :
        print(i, end=" ")
        break

    factor *= 10
    temp //= 10

  if i % 100 == 0 :
    print()
```

```python
# Print all numbers from 1 to 500 that are divisible by every non-zero digit they contain.
for i in range(1, 501) :
  temp, flag = i, True
  while temp > 0 :
    digit = temp % 10

    if digit and i % digit != 0 :
      flag = False
      break

    temp //= 10

  if flag:
    print(i, end=" ")

  if i % 100 == 0 :
    print()
```

```python
# Find the LCM and GCD of two numbers.
a = int(input("Enter the first number: "))
b = int(input("Enter the second number: "))
n1, n2 = a, b
while n2 > 0 :
  n1, n2 = n2, n1 % n2

gcd = n1
lcm = (a * b) // gcd
print("GCD: ",gcd)
print("LCM: ",lcm)
```
