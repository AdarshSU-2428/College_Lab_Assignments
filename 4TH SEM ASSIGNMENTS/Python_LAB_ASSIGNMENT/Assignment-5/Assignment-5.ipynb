```python
# 3. Count the number of items present in a tuple and then add two new items into it (one item using concatenation operator and another item using built-in function).
tup = (76, 56, 34, 75, 98)
print(len(tup))
tup1 = (54, 89)
tup2 = tup + tup1
print(tup2)
```

```python
# 4. Find the number of occurrences for any repeated item present in a tuple.
tup = (54, 45, 12, 89, 89, 77, 45)
l = []
for ele in tup :
    if ele not in l :
        print(ele,":",tup.count(ele))
        l.append(ele)
```

```python
# 7. Remove the odd number elements from a tuple.
tup = (35, 23, 56, 18, 42, 91)
for ele in tup :
    if ele % 2 != 0 :
        print(ele, end=" ")
```

```python
# 8. Reverse the elements of a tuple by using and without using a built-in function.
tup = (98, 34, 24, 67, 19, 58)
print("Sorted without in-built function: ",tup[::-1])
print("Sorted with in-built function: ",tuple(reversed(tup)))
```

```python
# 13. Given a tuple of numbers, create another tuple such that it contains only the even numbers of the first tuple.
tup = (45, 36, 27, 28, 18, 68)
t1 = tuple(ele for ele in tup if ele % 2 == 0)
print(tuple(sorted(t1)))
```

```python
# 14. Display the largest number of a tuple without using the built-in function max().The user need to input the values in tuple from the keyboard.
tup = (34, 87, 67, 98, 39)
res = tup[0]

for element in tup[1:]:
    if element > res:
        res = element

print("The largest element is:", res)
```

```python
# 15. Replace the last value of the tuple with 100. Sample list: [(1, 2, 3), (4, 5, 6), (7, 8, 9)] Output: [(1, 2, 100), (4, 5, 100), (7, 8, 100)]
list = [(1, 2, 3), (4, 5, 6), (7, 8, 9)]
l1 = []
for tup in list :
    l1.append(tup[:2] + (100,))

print(l1)
```

```python
# 16. Sort a tuple by its float element. Sample data: [('item1', '12.20'), ('item2', '15.10'), ('item3', '24.5')] Output: [('item3', '24.5'), ('item2', '15.10'), ('item1', '12.20')]
lst = [('item1', '12.20'), ('item2', '15.10'), ('item3', '24.5')]
l1 = []

for tup in lst:
    l1.append((tup[0], float(tup[1])))

for i in range(len(l1)):
    for j in range(i + 1, len(l1)):
        if l1[i][1] < l1[j][1]:
            l1[i], l1[j] = l1[j], l1[i]

print(l1)
```

```python
# 17. Count the number of elements in a tuple until an element is a list.
tup = (7, 8, 3, 9, [4, 8, 2, 9])
res = 0

for element in tup:
    if type(element) == list:
        break

    res += 1

print("Count is:", res)
```

```python
# 18. Swap the two tuples: tuple1 = (11, 22) tuple2 = (99, 88) Output: tuple1 = (99, 88) tuple2 = (11, 22)

tup1 = (11, 22)
tup2 = (99, 88)

if tup1 < tup2:
    tup1, tup2 = tup2, tup1

print("After sorting:", tup1, tup2)
```
