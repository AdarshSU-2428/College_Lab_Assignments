```python
# 17
df = pd.DataFrame({
'RollNo':[1,2,3,4],
'Name':['A','B','C','D'],
'Marks1':[80,70,90,60],
'Marks2':[85,75,95,65],
'Marks3':[90,80,85,70]
})

df['Total'] = df[['Marks1','Marks2','Marks3']].sum(axis=1)
df['Percentage'] = df['Total']/3

df['Grade'] = df['Percentage'].apply(lambda x: 'A' if x>=75 else 'B')

print(df[df['Percentage']>75])

print(df.sort_values('Percentage',ascending=False).iloc[1])

print(df.sort_values('Percentage'))
```

```python
# 18
df = pd.DataFrame({
'EmpID':[1,2,3,4],
'Name':['A','B','C','D'],
'Department':['HR','IT','IT','HR'],
'Salary':[20000,30000,40000,25000]
})

dept = input("Enter department: ")
print(df[df['Department']==dept])

df['Salary'] *= 1.10

print(df['Salary'].max(), df['Salary'].min())

print(df.groupby('Department')['Salary'].mean())

df.to_csv("employees.csv", index=False)
```

```python
# 19
df = pd.DataFrame({
'RollNo':[1,2,3],
'Name':['A','B','C'],
'Marks':[85,70,90],
'City':['Delhi','Mumbai','Delhi']
})

print(df[df['Marks']>80])

city = input("Enter city: ")
print(df[df['City']==city])

print(df.loc[:,['Name','Marks']])
print(df.iloc[:,1:3])

df.loc[df['Name']=='A','Marks'] = 95

df = df.drop(columns=['City'])

print(df)
```

```python
# 20
df = pd.DataFrame({
'ProductID':[1,2,3],
'ProductName':['A','B','C'],
'Category':['X','Y','X'],
'Price':[100,200,150],
'Quantity':[5,2,8]
})

df['Revenue'] = df['Price']*df['Quantity']

print(df['Revenue'].sum())

print(df.sort_values('Quantity',ascending=False))

print(df[df['Quantity']<3])

print(df.groupby('Category')['Revenue'].sum())

df.to_excel("sales.xlsx", index=False)
```
