```python
# 18. Arrange string characters such that lowercase letters should come first.
s = input("Enter the string: ")
lower = ""
upper = ""
for ch in s :
  if ch.islower() :
    lower += ch
  else :
    upper += ch

print(lower+upper)
```

```python
# 19. Count all lower case, upper case, digits, and special symbols from a given string.
s = input("Enter the string: ")
count_lower = 0
count_upper = 0
count_digit = 0
count_symbol = 0

for ch in s :
  if ch.islower() :
    count_lower += 1
  elif ch.isupper() :
    count_upper += 1
  elif ch.isdigit() :
    count_digit += 1
  else :
    count_symbol += 1

print("Lower case letters: ",count_lower)
print("Upper case letters: ",count_upper)
print("Digits: ",count_digit)
print("Special symbols: ",count_symbol)
```

```python
# 20. Given a string, return the sum and average of the digits that appear in the string, ignoring all other characters.
s = input("Enter a string: ")
digits =  0
count = 0
for ch in s :
  if ch.isdigit() :
    digits += int(ch)
    count += 1

avg = digits / count
print("Sum: ",digits)
print("Average: ",avg)
```

```python
# 21. Write a Python program to remove duplicates from a list
nums = input("Enter a number: ").split()
l = []
for n in nums :
  l.append(int(n))

print("The original list: ",l)

l1 = []
for i in l:
  if i not in l1 :
    l1.append(i)

print("The list after removed duplicates: ",l1)
```

```python
# 22. Write a Python program to find the second largest number in a list.
nums = input("Enter a number: ").split()
l = []
for n in nums :
  l.append(int(n))

print("The original list: ",l)

l1 = []

for i in l :
  if i not in l1 :
    l1.append(i)

l1.sort(reverse=True)
print("The reverse sorted unique list: ",l1)
print("The second largest number is: ",l1[1])
```

```python
# 23. Write a Python program to sort a list of elements using the bubble sort algorithm.
nums = input("Enter a number: ").split()
l = []
for n in nums :
  l.append(int(n))

print("The original list: ",l)

for i in range(len(l)-1) :
  for j in range(len(l)-i-1) :
    if l[j] > l[j+1] :
      l[j], l[j+1] = l[j+1], l[j]

print("The sorted list: ",l)
```

```python
# 25. Write a Python program to count the frequency of each element in a list.
nums = input("Enter a number: ").split()
l = []
for n in nums :
  l.append(int(n))

print("The original list: ",l)

l1 = []

for i in l :
  if i not in l1 :
    l1.append(i)

l1.sort()
print(l1)

print("Element frequency: ")
for items in l1 :
    count = 0
    for item in l:
        if item == items:
            count += 1
    print(items,":",count)
```

```python
# 26. Write a Python program to count the number of strings where the string length is 2 or more and the first and last character are same from a given list of strings
strings = input("Enter some strings: ").split()
l = []
for string in strings :
  l.append(string)

print(l)

count = 0
for s in l :
  if (len(s) >= 2) and (s[0] == s[-1]) :
    count += 1

print("No. of strings: ",count)

```
