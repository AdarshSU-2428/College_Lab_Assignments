```python
# 2. Create a class Rectangle with:
# • length
# • width
# Add methods:
# • area() → returns area
# • perimeter() → returns perimeter

class Rectangele:
    def __init__(self, length, width):
        self.length = length
        self.width = width
    def area(self) :
        return self.length * self.width
    def perimeter(self) :
        return 2 * (self.length + self.width)

rec = Rectangele(10, 20)
print("The area of the rectangle:",rec.area(),"m^2")
print("The perimeter of the rectangle:",rec.perimeter(),"m")
```

```python
# 4. Create a class Circle with attribute radius.
# Add methods:
# • area()
# • circumference()
# Use the value π = 3.14.

class Circle :
  def __init__(self, radius) :
    self.radius = radius
    self.pi = 3.14
  def area(self) :
    return self.pi * self.radius * self.radius
  def circumference(self) :
    return 2 * self.pi * self.radius

cir = Circle(10)
print("The area of the circle:",cir.area(),"m^2")
print("The circumference of the circle:",cir.circumference(),"m")
```

```python
# 5. Create a class BankAccount with:
# • account_holder
# • balance
# Methods:
# • deposit(amount)
# • withdraw(amount)
# • display_balance()
# Add a condition so withdrawal cannot exceed balance.

class BankAccount:
  def __init__(self, account_holder, balance=0):
    self.account_holder = account_holder
    self.balance = balance

  def deposit(self, amount):
      if amount > 0:
          self.balance += amount
          print("Deposited:", amount)
      else:
          print("Invalid deposit amount")

  def withdraw(self, amount):
      if amount > self.balance:
          print("Insufficient balance! Withdrawal denied.")
      elif amount <= 0:
          print("Invalid withdrawal amount")
      else:
          self.balance -= amount
          print("Withdrawn:", amount)

  def display_balance(self):
      print("Account Holder:", self.account_holder)
      print("Current Balance:", self.balance)

acc = BankAccount("Adarsh", 1000)
acc.deposit(500)
acc.withdraw(200)
acc.withdraw(2000)
acc.display_balance()
```

```python
# 8. Create a class Temperature with attribute celsius.
# Methods:
# • to_fahrenheit()
# • to_kelvin()

class Temperature:
    def __init__(self, celsius):
        self.celsius = celsius

    def to_fahrenheit(self):
        return (self.celsius * 9/5) + 32

    def to_kelvin(self):
        return self.celsius + 273.15

temp = Temperature(25)
print(temp.to_fahrenheit())
print(temp.to_kelvin())
```

```python
# 9. Create a class Library with:
# • book_list
# Methods:
# • add_book(book)
# • remove_book(book)
# • display_books()

class Library:
    def __init__(self):
        self.book_list = []

    def add_book(self, book):
        self.book_list.append(book)

    def remove_book(self, book):
        if book in self.book_list:
            self.book_list.remove(book)

    def display_books(self):
        for book in self.book_list:
            print(book)

lib = Library()
lib.add_book("Python Programming")
lib.add_book("Data Structures")
lib.display_books()
lib.remove_book("Python Programming")
lib.display_books()
```

```python
# 10. Create a class Cart with:
# • items (dictionary)
# Methods:
# • add_item(item, price)
# • remove_item(item)
# • total_price()

class Cart:
    def __init__(self):
        self.items = {}

    def add_item(self, item, price):
        self.items[item] = price

    def remove_item(self, item):
        if item in self.items:
            del self.items[item]

    def total_price(self):
        return sum(self.items.values())


cart = Cart()
cart.add_item("Laptop", 50000)
cart.add_item("Mouse", 500)
print(cart.total_price())
cart.remove_item("Mouse")
print(cart.total_price())
```

```python
# 11. Create a class Student with:
# • name
# • marks (list)
# Methods:
# • average_marks()
# • grade()
# • high_grad() Grades:
# • A → 90+
# • B → 75+
# • C → 50+
# • F → below 50

class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def average_marks(self):
        return sum(self.marks) / len(self.marks)

    def grade(self):
        avg = self.average_marks()
        if avg >= 90:
            return "A"
        elif avg >= 75:
            return "B"
        elif avg >= 50:
            return "C"
        else:
            return "F"

    def high_grad(self):
        return max(self.marks)

student = Student("Bob", [80, 90, 70])
print(student.average_marks())
print(student.grade())
print(student.high_grad())
```

```python
# 12. Create a class ATM with:
# • balance
# • pin
# Methods:
# • check_balance()
# • deposit()
# • withdraw()
# • change_pin()

class ATM:
    def __init__(self, balance, pin):
        self.balance = balance
        self.pin = pin

    def check_balance(self):
        return self.balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            return True
        else:
            return False

    def change_pin(self, new_pin):
        self.pin = new_pin

atm = ATM(10000, 1234)
print(atm.check_balance())
atm.deposit(2000)
print(atm.check_balance())
atm.withdraw(5000)
print(atm.check_balance())
atm.change_pin(4321)
print(atm.pin)
```

```python
# 13. Write a Python program to create a class representing a stack data structure. Include methods for pushing, popping and displaying elements.

class Stack:
    def __init__(self):
        self.stack = []

    def push(self, item):
        self.stack.append(item)

    def pop(self):
        if self.stack:
            return self.stack.pop()
        else:
            return None

    def display(self):
        print(self.stack)

stack = Stack()
stack.push(10)
stack.push(20)
stack.display()
print(stack.pop())
stack.display()
```

```python
# 14. Create a Person class with attributes like name and date_of_birth. Implement a method that calculates the person's current age based on the date of birth.
from datetime import date

class Person:
    def __init__(self, name, date_of_birth):
        self.name = name
        self.date_of_birth = date_of_birth  # format: YYYY-MM-DD

    def calculate_age(self):
        dob = date.fromisoformat(self.date_of_birth)
        today = date.today()
        age = today.year - dob.year
        if (today.month, today.day) < (dob.month, dob.day):
            age -= 1
        return age

person = Person("Charlie", "2000-05-15")
print("The person's age: ",person.calculate_age())
```

```python
# 16. Define a class Point, which can be used to create two-dimensional points. Create three objects of Point class and check, these points are non collinear.
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

def are_non_collinear(p1, p2, p3):
    area = p1.x * (p2.y - p3.y) + p2.x * (p3.y - p1.y) + p3.x * (p1.y - p2.y)
    return area != 0

p1 = Point(0, 0)
p2 = Point(1, 1)
p3 = Point(2, 3)

print(are_non_collinear(p1, p2, p3))
```

```python
# 18. Define a class ComplexNumber, placed the necessary data members and member functions to perform basic Complex number operation.
class ComplexNumber:
    def __init__(self, real=0, imag=0):
        self.real = real
        self.imag = imag

    def add(self, other):
        return ComplexNumber(self.real + other.real, self.imag + other.imag)

    def subtract(self, other):
        return ComplexNumber(self.real - other.real, self.imag - other.imag)

    def multiply(self, other):
        real_part = self.real * other.real - self.imag * other.imag
        imag_part = self.real * other.imag + self.imag * other.real
        return ComplexNumber(real_part, imag_part)

    def display(self):
        print(self.real, "+", self.imag, "i")

c1 = ComplexNumber(3, 2)
c2 = ComplexNumber(1, 7)

c3 = c1.add(c2)
c3.display()

c4 = c1.subtract(c2)
c4.display()

c5 = c1.multiply(c2)
c5.display()
```

```python
# 19. Define a class Time having data members hour, minute and second. The class should have a parameterized constructor to initialize its data members. It should also have methods displayTime() to display the time in HH:MM: SS format and addTime() to add two-time objects.
class Time:
    def __init__(self, hour, minute, second):
        self.hour = hour
        self.minute = minute
        self.second = second

    def displayTime(self):
        print(f"{self.hour:02d}:{self.minute:02d}:{self.second:02d}")

    def addTime(self, other):
        total_seconds = self.second + other.second
        total_minutes = self.minute + other.minute + total_seconds // 60
        total_hours = self.hour + other.hour + total_minutes // 60

        new_second = total_seconds % 60
        new_minute = total_minutes % 60
        new_hour = total_hours % 24

        return Time(new_hour, new_minute, new_second)

t1 = Time(2, 45, 50)
t2 = Time(3, 20, 30)

t1.displayTime()
t2.displayTime()

t3 = t1.addTime(t2)
t3.displayTime()
```

```python
# Q21. Define a class Student as follows.
# •	Data Members
#   •	studName
#   •	studRoll
#   •	studMarks[5]
#   •	studBranch
#   •	studCollegeName
# •	Member Functions
#   •	dispStudentDetails()
# Create an array of 10 student object. Display the details information about the student who has secured highest average mark.

class Student:
    def __init__(self, studName, studRoll, studMarks, studBranch, studCollegeName):
        self.studName = studName
        self.studRoll = studRoll
        self.studMarks = studMarks
        self.studBranch = studBranch
        self.studCollegeName = studCollegeName

    def averageMarks(self):
        return sum(self.studMarks) / len(self.studMarks)

    def dispStudentDetails(self):
        print("Name:", self.studName)
        print("Roll:", self.studRoll)
        print("Branch:", self.studBranch)
        print("College:", self.studCollegeName)
        print("Marks:", self.studMarks)
        print("Average:", self.averageMarks())


students = [
    Student("Raman", 1, [80, 90, 85, 70, 95], "CSE", "IIT"),
    Student("Sita", 2, [60, 75, 70, 65, 80], "ECE", "NIT"),
    Student("Ajay", 3, [90, 92, 88, 85, 91], "CSE", "VIT"),
    Student("Ayushi", 4, [50, 55, 60, 58, 62], "MechE", "NIT"),
    Student("Kamal", 5, [70, 72, 68, 75, 74], "CSE", "IIT"),
    Student("Priya", 6, [85, 90, 88, 92, 89], "EEE", "NIT")
]

highest_student = students[0]
for s in students:
    if s.averageMarks() > highest_student.averageMarks():
        highest_student = s

highest_student.dispStudentDetails()
```

```python
# 22. Develop a Python program that will deal with employee information of an organization. Define a class Employee, which has the data members and member functions are follows.
# •	Data Members
#   •	empName
#   •	empNumber
#   •	basicSal
#   •	DA
#   •	HRA
# •	Member Functions
#   •	calGrossSal ( )
#   •	showEmpDetails( )
# Create an array of objects. Display those employees, whose gross salary is more than 50,000.
class Employee:
    def __init__(self, empName, empNumber, basicSal, DA, HRA):
        self.empName = empName
        self.empNumber = empNumber
        self.basicSal = basicSal
        self.DA = DA
        self.HRA = HRA

    def calGrossSal(self):
        return self.basicSal + self.DA + self.HRA

    def showEmpDetails(self):
        print("Employee Name:", self.empName)
        print("Employee Number:", self.empNumber)
        print("Basic Salary:", self.basicSal)
        print("DA:", self.DA)
        print("HRA:", self.HRA)
        print("Gross Salary:", self.calGrossSal())
        print()


employees = [
    Employee("Ravi Kumar", 101, 30000, 15000, 8000),
    Employee("Sneha Patnaik", 102, 40000, 12000, 10000),
    Employee("Amit Sharma", 103, 25000, 10000, 7000),
    Employee("Priya Das", 104, 35000, 18000, 12000),
    Employee("Arjun Mehta", 105, 28000, 9000, 6000),
    Employee("Kiran Rao", 107, 20000, 8000, 5000),
]

print("Employees with Gross Salary > 50000:")
for emp in employees:
    if emp.calGrossSal() > 50000:
        emp.showEmpDetails()
```
