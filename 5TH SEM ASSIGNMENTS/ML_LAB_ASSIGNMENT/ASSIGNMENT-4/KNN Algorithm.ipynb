```python
# Import necessary libraries:
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import math
from collections import Counter
```

```python
# Load the dataset from CSV file into a pandas DataFrame
data = pd.read_csv("knn_dataset.csv")

# Display the dataset table
data
```

```python
# Separate features (X1, X2, X3) and target labels (Y):
train_data = data.iloc[:, 1:-1].values

# y_train: extracts the target class labels from the last column ('Y')
y_train = data.iloc[:, -1].values
```

```python
# Display the extracted training feature matrix
train_data
```

```python
# Display the target class labels for all training points
y_train
```

```python
# Function to calculate Euclidean distance between two points in n-dimensional space
def euclidean_distance(p1, p2):
     return math.sqrt(sum((a - b) ** 2 for a, b in zip(p1, p2)))  # Formula: d(p1, p2) = sqrt(sum((a - b)^2))

# KNN classifier function:
def knn_classify(train_data, test_point, k):
    distances = []
    
    # Calculate Euclidean distance to each point in the training set
    for i in range(len(train_data)):
        dist = euclidean_distance(test_point, train_data[i])
        print("The euclidean distances: ", dist)
        distances.append((dist, y_train[i]))
        
    distances.sort(key=lambda x:x[0])   # Sort distances in ascending order (nearest neighbors first)
    
    k_nearest = [label for(_, label) in distances[:k]]   # Extract labels of the top k nearest neighbors
    
    most_common = Counter(k_nearest).most_common(1)   # Find the most common label among the k neighbors (majority vote)
    
    return most_common[0][0]   # Return the predicted class label
```

```python
# Define the query / test point coordinates [X1, X2, X3] to classify
test_point = [0, 0, 0]
print("The test point: ", test_point)

# Set the value of k (number of nearest neighbors)
k = 3
print("The k value: ", k)
```

```python
# Run the KNN classifier on the test point with k=3
pred_label = knn_classify(train_data, test_point, k)
```

```python
# Display the final predicted class label
print("The predicted label: ", pred_label)
```
