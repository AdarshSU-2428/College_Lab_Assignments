```python
#Importing Libraries
import numpy as np
import pandas as pd
```

```python
#Ridge Regression Algorithm function (beta_ridge = inverse(X^T * X + lambda * I) * X^T * y))
def ridge_regression(X, y, lm=None):
    X_mean = np.mean(X, axis=0)
    X_norm = X - X_mean

    y_mean = np.mean(y)
    y_norm = y - y_mean

    XTX = X_norm.T @ X_norm
    I = np.eye(X.shape[1])

    beta_ridge = np.linalg.inv(XTX + lm * I) @ (X_norm.T @ y_norm)
    intercept = y_mean - np.dot(X_mean, beta_ridge)

    return beta_ridge, intercept
```

```python
#Random Dataset
data = {
    'X1': [1, 2, 3],
    'X2': [2, 4, 2],
    'Y':  [1, 2, 3]
}

df = pd.DataFrame(data)
```

```python
#Splitting dataset 
X = df[['X1', 'X2']].to_numpy(dtype=float)
y = df['Y'].to_numpy(dtype=float)
```

```python
# Train the model
beta, intercept = ridge_regression(X, y, lm=1.0)

print("Coefficients:", beta)
print("Intercept:", intercept)

# Prediction
X_new = np.array([[6, 7]])
y_pred = X_new @ beta + intercept

print("Predicted value:", y_pred[0])
```
