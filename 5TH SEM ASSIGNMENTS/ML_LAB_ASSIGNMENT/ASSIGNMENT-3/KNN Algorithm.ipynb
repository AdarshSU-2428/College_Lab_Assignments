```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import math
from collections import Counter
```

```python
data = pd.read_csv("knn_dataset.csv")
data
```

```python
train_data = data.iloc[:, 1:-1].values
y_train = data.iloc[:, -1].values
```

```python
train_data
```

```python
y_train
```

```python
def euclidean_distance(p1, p2):
     return math.sqrt(sum((a - b) ** 2 for a, b in zip(p1, p2)))

def knn_classify(train_data, test_point, k):
    distances = []
    
    for i in range(len(train_data)):
        dist = euclidean_distance(test_point, train_data[i])
        print("The euclidean distances: ", dist)
        distances.append((dist, y_train[i]))
        
    distances.sort(key=lambda x:x[0])
    k_nearest = [label for(_, label) in distances[:k]]
    
    most_common = Counter(k_nearest).most_common(1)
    
    return most_common[0][0]
```

```python
test_point = [0, 0, 0]
print("The test point: ", test_point)

k = 3
print("The k value: ", k)
```

```python
pred_label = knn_classify(train_data, test_point, k)
```

```python
print("The predicted label: ", pred_label)
```
