```python
# Import required data analysis libraries
import pandas as pd
import numpy as np
```

```python
# Load weather dataset from CSV file
csv_path = 'temperature.csv'
df = pd.read_csv(csv_path)
```

```python
# Inspect the full dataset
print("=== Weather Dataset Overview ===")
print(df)
```

```python
# Construct DataFrame manually from dictionary of records
custom_weather_data = {
    'day': ['1/1/2017', '1/2/2017', '1/3/2017', '1/4/2017', '1/5/2017', '1/6/2017'],
    'temperature': [32, 35, 28, 24, 32, 31],
    'windspeed': [6, 7, 2, 7, 4, 2],
    'humidity': [20, 20, 20, 20, 20, 20],
    'event': ['Rain', 'Sunny', 'Snow', 'Snow', 'Rain', 'Sunny']
}

new_df = pd.DataFrame(custom_weather_data)
print("=== Manually Created DataFrame ===")
print(new_df)
```

```python
# View column labels and schema
print("Columns in DataFrame:")
print(list(df.columns))
```

```python
# View row index range
print("Row Index Range:")
print(df.index)
```

```python
# Preview initial 5 records (default head)
print("First 5 rows of data:")
print(df.head(n=5))
```

```python
# Preview first 2 records
print("First 2 rows of data:")
print(df.head(n=2))
```

```python
# Preview trailing 5 records (default tail)
print("Last 5 rows of data:")
print(df.tail(n=5))
```

```python
# Preview trailing 2 records
print("Last 2 rows of data:")
print(df.tail(n=2))
```

```python
# Sample 3 random rows from the DataFrame
print("Random sample (3 records):")
print(df.sample(n=3))
```

```python
# Extract and display all column names as an array
all_columns = df.columns.values
print("All Column Names:", all_columns)
```

```python
# Extract a single column as a Series
print("Selecting single column ['day']:")
print(df['day'])
```

```python
# Select a subset of multiple columns
selected_features = ['day', 'temperature']
print("Selected columns (day & temperature):")
print(df[selected_features])
```

```python
# Generate summary statistics for numeric features
print("=== Descriptive Statistical Summary ===")
print(df.describe())
```

```python
# Summary statistics by column data type
print("--- Numerical Feature Analysis (temperature) ---")
print(df['temperature'].describe())

print("--- Categorical Feature Analysis (day) ---")
print(df['day'].describe())
```

```python
# DataFrame structural info (dtypes, non-null counts, memory)
print("=== DataFrame Info & Schema ===")
df.info()
```

```python
# Indexing and Slicing with .iloc and .loc
print("Full DataFrame:")
print(df)

# Position-based access 
print("[iloc] Value at Row 2, Col 4:", df.iloc[2, 4])

# Label/Index-based row slicing 
print("[loc] Rows with index labels 2 and 3:")
print(df.loc[[2, 3]])

# Label-based scalar lookup 
print("[loc] Temperature at Row 2:", df.loc[2, 'temperature'])
```

```python
# Compute key aggregate metrics for temperature
temp_series = df['temperature']
print(f"Max Temperature : {temp_series.max()}")
print(f"Min Temperature : {temp_series.min()}")
print(f"Std Deviation   : {temp_series.std():.4f}")
print(f"Mean Temperature: {temp_series.mean():.2f}")
```

```python
# Create a boolean condition for max temperature
is_max_temp = df['temperature'] == df['temperature'].max()
print("Boolean condition for max temperature:")
print(is_max_temp)
```

```python
# Filter DataFrame to extract row(s) with highest temperature
max_temp_record = df[df['temperature'] == df['temperature'].max()]
print("Record with maximum temperature:")
print(max_temp_record)
```

```python
# Extract the specific 'day' value corresponding to max temperature
hottest_day = df.loc[df['temperature'] == df['temperature'].max(), 'day']
print("Date with maximum temperature:")
print(hottest_day.to_string(index=False))
```

```python
# Display current state of the DataFrame
print("=== Current DataFrame State ===")
print(df)
```

```python
# Re-indexing by setting 'day' as index 
df_indexed = df.set_index('day')
print("=== DataFrame Re-indexed by 'day' ===")
print(df_indexed)

print("--- Querying by index label '1/1/2017' ---")
print(df_indexed.loc['1/1/2017'])

print("--- Original DataFrame (Index Unchanged) ---")
print(df)
```

```python
# Filter rows matching a specific date ('1/4/2017')
target_date_record = df[df['day'] == '1/4/2017']
print("Record for date '1/4/2017':")
print(target_date_record)
```

```python
# Sort DataFrame based on 'windspeed' in ascending order
sorted_df = df.sort_values(by=['windspeed'], ascending=True)
print("=== DataFrame Sorted by Windspeed ===")
print(sorted_df)
```
