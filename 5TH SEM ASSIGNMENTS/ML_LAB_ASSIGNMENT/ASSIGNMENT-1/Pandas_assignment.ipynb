```python
# Setup imports and initial data
import numpy as np
import pandas as pd

data = {
    'birds': ['Cranes', 'Cranes', 'plovers', 'spoonbills', 'spoonbills', 'Cranes', 'plovers', 'Cranes', 'spoonbills', 'spoonbills'], 
    'age': [3.5, 4, 1.5, np.nan, 6, 3, 5.5, np.nan, 8, 4],
    'visits': [2, 4, 3, 4, 3, 4, 2, 2, 3, 2],
    'priority': ['yes', 'yes', 'no', 'yes', 'no', 'no', 'no', 'yes', 'no', 'no']
}

labels = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j']
```

```python
# 1. Create a DataFrame birds from this dictionary data which has the index labels.
birds = pd.DataFrame(data, index=labels)

birds
```

```python
# 2. Display a summary of the basic information about birds DataFrame and its data.
birds.info()
birds.describe()
```

```python
# 3. Print the first 2 rows of the birds dataframe
birds.head(2)
```

```python
# 4. Print all the rows with only 'birds' and 'age' columns from the dataframe
birds[['birds', 'age']]
```

```python
# 5. select [2, 3, 7] rows and in columns ['birds', 'age', 'visits']
birds.iloc[[2, 3, 7]][['birds', 'age', 'visits']]
```

```python
# 6. select the rows where the number of visits is less than 4
birds[birds['visits'] < 4]
```

```python
# 7. select the rows with columns ['birds', 'visits'] where the age is missing i.e NaN
birds[birds['age'].isna()][['birds', 'visits']]
```

```python
# 8. Select the rows where the birds is a Cranes and the age is less than 4
birds[(birds['birds'] == 'Cranes') & (birds['age'] < 4)]
```

```python
# 9. Select the rows the age is between 2 and 4(inclusive)
birds[birds['age'].between(2, 4)]
```

```python
# 10. Find the total number of visits of the bird Cranes
birds[birds['birds'] == 'Cranes']['visits'].sum()
```

```python
# 11. Calculate the mean age for each different birds in dataframe.
birds.groupby('birds')['age'].mean()
```

```python
# 12. Append a new row 'k' to dataframe with your choice of values for each column. Then delete that row to return the original DataFrame.
# Append row 'k'
birds.loc['k'] = ['sparrow', 2.0, 1, 'yes']
print("After appending row 'k':\n", birds)

# Delete row 'k'
birds = birds.drop('k')
print("\nAfter deleting row 'k':\n", birds)
```

```python
# 13. Find the number of each type of birds in dataframe (Counts)
birds['birds'].value_counts()
```

```python
# 14. Sort dataframe (birds) first by the values in the 'age' in decending order, then by the value in the 'visits' column in ascending order.
birds.sort_values('visits').sort_values('age', ascending=False)
```

```python
# 15. Replace the priority column values with'yes' should be 1 and 'no' should be 0
birds['priority'] = birds['priority'].map({'yes': 1, 'no': 0})

birds
```

```python
# 16. In the 'birds' column, change the 'Cranes' entries to 'trumpeters'.
birds['birds'] = birds['birds'].replace('Cranes', 'trumpeters')

birds
```
